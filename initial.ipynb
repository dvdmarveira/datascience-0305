```python
data_int = pd.date_range("20250101", periods=6, freq="ME")
```

```python
data_br = data_int.strftime("%d/%m/%Y")
data_br
```

```python
df= pd.DataFrame(
    np.random.randint(1,5,size=(6,4)),
    index=data_br,
    columns=["A", "B", "C", "D"])
```

```python
df.head(3)
```

```python
df.columns = ['a', 'b', 'c', 'd']
```

```python
df['Col5'] = pd.Series([1,2,3,4,5,6], index=data_br)
```
